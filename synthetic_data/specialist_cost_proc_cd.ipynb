```python
# Generate a synthetic dataset with 12,000 rows including cost variation, outliers, and risk scoring
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)

n = 12000

specialties = ["Cardiology","Family Practice","Orthopedics","Internal Medicine","Endocrinology","Dermatology"]
submarkets = ["EAST TEXAS","CENTRAL TX","EASTSIDE","SOUTH TX","LAREDO"]
plan_types = ["MAPD"]

entities = {
    "RENAISSANCE (RPO)": ["RPO EASTEX","RPO CENTRAL","RPO EASTSIDE"],
    "VALLEY ORGANIZED PHYSICIANS LLC (VOP)": ["DOCTORS ORGANIZE CARE","ALLIANCE OF LAREDO"]
}

procedure_codes = [
    ("99213","Office Visit",80,140),
    ("99214","Office Visit Extended",100,180),
    ("93000","ECG Routine",70,130),
    ("73562","Knee X-ray",120,220),
    ("84443","TSH Test",30,80),
    ("11102","Skin Biopsy",150,300),
    ("92928","Coronary Stent",5000,8000)
]

dx_codes = [
    ("I10","Hypertension"),
    ("E11.9","Type 2 Diabetes"),
    ("M17.9","Knee Osteoarthritis"),
    ("J06.9","URI"),
    ("I25.10","CAD"),
    ("E03.9","Hypothyroidism"),
    ("L40.9","Psoriasis")
]

rows = []

start_date = datetime(2025,1,1)

for i in range(n):
    entity = random.choice(list(entities.keys()))
    pod = random.choice(entities[entity])
    proc = random.choice(procedure_codes)
    dx = random.choice(dx_codes)
    
    base_paid = random.uniform(proc[2], proc[3])
    
    # Introduce outliers (~2%)
    if random.random() < 0.02:
        base_paid *= random.uniform(2,5)
    
    total_allowed = base_paid * random.uniform(1.1,1.4)
    
    # Risk score (0-1 scaled)
    risk_score = round(min(1, (base_paid / 8000) + random.uniform(0,0.2)),3)
    
    rows.append({
        "Specialty": random.choice(specialties),
        "OperationalMarket": "TX",
        "OperationalSubMarket": random.choice(submarkets),
        "PlanType": "HMO",
        "ManagingEntity": entity,
        "ReportingPod": pod,
        "PodName": f"Pod {random.choice(['A','B','C','D','E'])}",
        "PCPName": f"Dr. {random.choice(['Smith','Lopez','Patel','Hernandez','Ruiz'])}",
        "MemberID": f"M{10000+i}",
        "memberFirst": random.choice(["John","Maria","Carlos","Ana","Linda","James","Sofia"]),
        "memberLast": random.choice(["Smith","Garcia","Brown","Martinez","Lopez","Nguyen"]),
        "gender": random.choice(["M","F"]),
        "ClaimID": f"C{90000+i}",
        "PrimaryDx": dx[0],
        "PrimaryDxDesc": dx[1],
        "ProcedureCode": proc[0],
        "ProcedureDesc": proc[1],
        "DOSbegin": start_date + timedelta(days=random.randint(0,365)),
        "TotalPaid": round(base_paid,2),
        "TotalAllowed": round(total_allowed,2),
        "RiskScore": risk_score,
        "ProviderID": f"P{5000+i}",
        "ProviderNPI": str(1000000000 + i),
        "ProviderName": f"Dr. {random.choice(['Allen','Chen','White','Gomez','Ruiz'])}",
        "ProvBillEntityName": random.choice(["East TX Group","Central Clinic","Valley Med","Laredo Care"]),
        "NetworkStatus": random.choice(["INN","OON"]),
        "ReferringProviderID": f"P{4000+i}" if random.random() > 0.3 else "",
        "RefferingProviderName": random.choice(["Dr. Smith","Dr. Lopez","Dr. Patel"]) if random.random() > 0.3 else "",
        "PlaceOfService": random.choice(["11","21","22","81"]),
        "PlaceofServiceDesc": random.choice(["Office","Inpatient","Outpatient","Lab"])
    })

df = pd.DataFrame(rows)

# Save files
csv_path = "/Users/bassel_instructor/Documents/datasets/medicare_synthetic_12k.csv"
#xlsx_path = "/mnt/data/medicare_synthetic_12k.xlsx"

df.to_csv(csv_path, index=False)
#df.to_excel(xlsx_path, index=False)

#csv_path, xlsx_path
```
